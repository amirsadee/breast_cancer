```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import tensorflow as tf 
from tensorflow import keras
from keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('/Users/air/desktop/scikit_learn/breast-cancer.csv')
```

```python
df
```

```python
df.drop('id',axis=1,inplace=True)
```

```python
print(df.shape)
```

```python
print(df.isnull().sum())
```

```python
df['diagnosis']=df['diagnosis'].replace(to_replace=['M','B'],value=[1,0])
```

```python
print(df)
```

```python
x = df.drop('diagnosis',axis=1)
y = df['diagnosis']
```

```python
x_train ,x_test , y_train ,y_test = train_test_split(x,y,test_size=0.20,random_state=False,shuffle=False)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train_scaled = scaler.transform(x_train)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(30,activation='relu'),
    tf.keras.layers.Dense(60,activation='relu'),
    tf.keras.layers.Dense(120,activation='relu'),
    tf.keras.layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(
    loss=tf.keras.losses.BinaryCrossentropy(),
    optimizer=tf.keras.optimizers.Adam(),
    metrics=[
         tf.keras.metrics.BinaryAccuracy(name='accuracy'),
         tf.keras.metrics.Precision(name='precision'),
         tf.keras.metrics.Recall(name='recall'),
     ]
 )
```

```python
history = model.fit(x_train_scaled,y_train,epochs=100)
```

```python
plt.figure(figsize=(18,6))
plt.plot(np.arange(1,101),history.history['accuracy'],label='accuracy')
plt.xlabel('epochs')
plt.ylabel('metrics')
plt.legend()
plt.show()
```
